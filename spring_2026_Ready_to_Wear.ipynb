# **primeiros passos**

importar bibliotecas

```python
# ========= Imports =========
import time
import re
import unicodedata
import random
import requests
import pandas as pd
import matplotlib.pyplot as plt
from bs4 import BeautifulSoup
from collections import Counter
from sklearn.feature_extraction.text import CountVectorizer

import spacy
from spacy.util import is_package
import spacy.cli
from spacy.tokenizer import Tokenizer

# ========= spaCy (baixa se necessário) =========
if not is_package("en_core_web_sm"):
    spacy.cli.download("en_core_web_sm")
nlp = spacy.load("en_core_web_sm")

# Tokenizer preservando hífens
infixes = [x for x in nlp.Defaults.infixes if "-" not in x]
infix_re = spacy.util.compile_infix_regex(infixes)
nlp.tokenizer = Tokenizer(
    nlp.vocab,
    rules=nlp.Defaults.tokenizer_exceptions,
    prefix_search=nlp.tokenizer.prefix_search,
    suffix_search=nlp.tokenizer.suffix_search,
    infix_finditer=infix_re.finditer,
    token_match=nlp.tokenizer.token_match,
)
```

**retrieving data**

```python
# ========= Passo 1: coletar links do índice =========
INDEX_URL = "https://www.vogue.com/fashion-shows/spring-2026-ready-to-wear"
HEADERS = {"User-Agent": "Mozilla/5.0"}

resp = requests.get(INDEX_URL, headers=HEADERS, timeout=30)
resp.raise_for_status()
soup_idx = BeautifulSoup(resp.text, "html.parser")

# seletor estável do menu lateral
links = []
for a in soup_idx.select('a[data-testid="navigation__internal-link"]'):
    href = a.get("href")
    if href and "/fashion-shows/" in href:
        links.append(href)

# remove duplicatas e mantém ordem
seen = set()
list_fashion_shows = []
for href in links:
    if href not in seen:
        seen.add(href)
        list_fashion_shows.append(href)

print(f"{len(list_fashion_shows)} links encontrados")
print(list_fashion_shows[:5])
```

```python
# ========= Passo 2: função para extrair review =========
def extract_vogue_review(url):
    """
    Extrai o texto completo do review de um desfile da Vogue.
    """
    r = requests.get(url, headers=HEADERS, timeout=20)
    r.raise_for_status()
    soup = BeautifulSoup(r.text, "html.parser")

    container = (
        soup.select_one('[data-journey-hook="runway-show-review-section"]')
        or soup.select_one('[data-testid="BodyWrapper"]')
        or soup.select_one("article")
    )

    if not container:
        return "", {"found_container": False, "paragraphs": 0, "length_chars": 0}

    # remove elementos inúteis (ex: botão "more")
    for tag in container.select("button, script, style"):
        tag.decompose()

    paragraphs = [p.get_text(" ", strip=True) for p in container.select("p")]
    full_text = "\n\n".join([p for p in paragraphs if p])

    debug = {
        "found_container": True,
        "paragraphs": len(paragraphs),
        "length_chars": len(full_text),
    }
    return full_text, debug

# ========= Passo 3: coleta em lote =========
BASE = "https://www.vogue.com"

articles = []          # só textos
articles_data = []     # textos + metadados

for i, rel in enumerate(list_fashion_shows, start=1):
    full_url = BASE + rel
    try:
        text, info = extract_vogue_review(full_url)
        if text and info["length_chars"] > 300:
            articles.append(text)
            articles_data.append({
                "url": full_url,
                "text": text,
                "paragraphs": info["paragraphs"],
                "length_chars": info["length_chars"],
            })
            print(f"✅ [{i}/{len(list_fashion_shows)}] {full_url} ({info['length_chars']} chars)")
        else:
            print(f"⚠️ [{i}/{len(list_fashion_shows)}] Vazio ou curto: {full_url}")
        time.sleep(1)  # respeita o servidor
    except Exception as e:
        print(f"❌ [{i}/{len(list_fashion_shows)}] Erro em {full_url}: {type(e).__name__} - {e}")
        time.sleep(2)

print(f"\nTotal de artigos coletados: {len(articles)} / {len(list_fashion_shows)}")

# Salvar CSV e construir corpus a partir do DF (mais seguro)
df_articles = pd.DataFrame(articles_data)
df_articles.to_csv("vogue_spring2026_reviews.csv", index=False)
print("✅ Arquivo salvo: vogue_spring2026_reviews.csv")


articles = df_articles["text"].fillna("").astype(str).tolist()
```

# **pré-processsamento de texto**

pré-processamento de texto

```python
# ========= Passo 4: pré-processamento =========
def clean_text(text):
    text = unicodedata.normalize("NFKD", text).encode("ascii", "ignore").decode("utf-8", "ignore")
    text = text.lower()
    text = re.sub(r"\d+", "", text)
    text = re.sub(r"[^\w\s-]", "", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

stopwords_spacy = nlp.Defaults.stop_words

def lemmatize_article(text):
    doc = nlp(text)
    return [
        token.lemma_ if token.lemma_ != "-PRON-" else token.text
        for token in doc
        if token.text.lower() not in stopwords_spacy and not token.is_punct
    ]

def combine_lemmas(preprocessed_articles):
    return [" ".join(article) for article in preprocessed_articles]

def preprocess_articles(articles):
    cleaned_articles = [clean_text(article) for article in articles]
    lemmatized_articles = [lemmatize_article(article) for article in cleaned_articles]
    combined_articles = combine_lemmas(lemmatized_articles)
    return combined_articles

preprocessed_articles_list = preprocess_articles(articles)       # p/ n-gramas
preprocessed_articles_str  = " ".join(preprocessed_articles_list)  # p/ buscas

print("Qtd artigos raspados:", len(articles))
print("Qtd docs preproc:", len(preprocessed_articles_list))
vazios = [i for i, t in enumerate(preprocessed_articles_list) if not t.strip()]
print("Docs vazios:", len(vazios), "->", vazios[:10])


```

# **identificando as tendências de moda**

n-gramas

```python
# ========= Passo 6: n-gramas =========
def generate_ngrams(corpus, n, top_k=20, filter_fashion=False):
    corpus = [c for c in corpus if c.strip()]
    if not corpus:
        return []

    vectorizer = CountVectorizer(
        ngram_range=(n, n),
        tokenizer=lambda text: text.split(),
        preprocessor=lambda text: text
    )
    X = vectorizer.fit_transform(corpus)
    if X.shape[1] == 0:
        return []

    ngram_counts = Counter(dict(zip(vectorizer.get_feature_names_out(), X.sum(axis=0).A1)))

    if filter_fashion:
        fashion_set = set(word for category in fashion_categories.values() for word in category)
        ngram_counts = {
            ngram: count
            for ngram, count in ngram_counts.items()
            if any(word in fashion_set for word in ngram.split())
        }

    return Counter(ngram_counts).most_common(top_k)

def plot_ngrams_color(ngrams, title, color):
    if not ngrams:
        print(f"(sem dados) {title}")
        return
    df = pd.DataFrame(ngrams, columns=["N-Grama", "Frequência"])
    plt.figure(figsize=(10, 6))
    plt.barh(df["N-Grama"], df["Frequência"], color=color)
    plt.xlabel("Frequência", fontsize=14)
    plt.ylabel("N-Gramas", fontsize=14)
    plt.title(title, fontsize=16)
    plt.gca().invert_yaxis()
    plt.show()

general_bigrams   = generate_ngrams(preprocessed_articles_list, 2, top_k=20)
fashion_bigrams   = generate_ngrams(preprocessed_articles_list, 2, top_k=20, filter_fashion=True)
general_trigrams  = generate_ngrams(preprocessed_articles_list, 3, top_k=30)
fashion_trigrams  = generate_ngrams(preprocessed_articles_list, 3, top_k=30, filter_fashion=True)

plot_ngrams_color(general_bigrams,  "Top 20 Bigramas Gerais", "#C0FF00")
plot_ngrams_color(fashion_bigrams,  "Top 20 Bigramas de Moda", "#FF007F")
plot_ngrams_color(fashion_trigrams, "Top 30 Trigramas de Moda", "#FF007F")
plot_ngrams_color(general_trigrams, "Top 30 Trigramas Gerais", "#C0FF00")
```

```python
# ========= Passo 7: nuvens =========
def color_function(word, font_size, position, orientation, random_state=None, **kwargs):
    return random.choice(["#FF007F", "#C0FF00"])

def criar_nuvem_moda(corpus, fashion_categories):
    todas_palavras = " ".join(corpus).split()
    termos_moda = set(word for categoria in fashion_categories.values() for word in categoria)
    palavras_moda = [word for word in todas_palavras if word in termos_moda]
    contagem = Counter(palavras_moda)
    contagem = dict(sorted(contagem.items(), key=lambda item: item[1], reverse=True))
    if not contagem:
        print("⚠️ Nenhum termo de moda encontrado no corpus.")
        return
    print("\n🔝 Top 10 palavras mais frequentes:")
    for i, (word, freq) in enumerate(list(contagem.items())[:10], 1):
        print(f"{i}. {word}: {freq}")
    from wordcloud import WordCloud
    wordcloud = WordCloud(width=800, height=400, background_color="white",
                          max_words=100, random_state=42,
                          prefer_horizontal=1.0, relative_scaling=1,
                          color_func=color_function).generate_from_frequencies(contagem)
    plt.figure(figsize=(10, 5))
    plt.imshow(wordcloud, interpolation="bilinear"); plt.axis("off")
    plt.title("Principais palavras da temporada Fall 2025 Ready-To-Wear", fontsize=14, fontweight="bold")
    plt.show()

criar_nuvem_moda(preprocessed_articles_list, fashion_categories)
```

```python
# ========= Passo 8: bigramas para ranking por categoria =========
def gerar_bigramas(corpus):
    bigramas = []
    for texto in corpus:
        palavras = texto.split()
        bigramas.extend([" ".join(pair) for pair in zip(palavras, palavras[1:])])
    return bigramas

bigramas_corpus = gerar_bigramas(preprocessed_articles_list)

def rank_fashion_categories(fashion_categories, bigrams_corpus):
    category_counts = {category: 0 for category in fashion_categories.keys()}
    for category, words in fashion_categories.items():
        wset = set(words)
        for bigram in bigrams_corpus:
            if any(word in wset for word in bigram.split()):
                category_counts[category] += 1
    df_categories = pd.DataFrame(list(category_counts.items()), columns=["Categoria", "Frequência"])
    return df_categories.sort_values(by="Frequência", ascending=False)

df_category_ranking = rank_fashion_categories(fashion_categories, bigramas_corpus)
display(df_category_ranking)
```
