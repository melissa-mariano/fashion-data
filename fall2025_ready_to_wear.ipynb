# **primeiros passos**

importar bibliotecas

```python
import requests
from bs4 import BeautifulSoup
from wordcloud import WordCloud
import matplotlib.pyplot as plt
from collections import Counter
from sklearn.feature_extraction.text import CountVectorizer
import spacy
import re
import unicodedata
import pandas as pd
import random

# carregar modelo spaCy
!python -m spacy download en_core_web_sm
nlp = spacy.load("en_core_web_sm")

# ajuste do tokenizer para preservar hífens
infixes = nlp.Defaults.infixes
infixes = [pattern for pattern in infixes if "-" not in pattern]
infix_re = spacy.util.compile_infix_regex(infixes)
nlp.tokenizer.infix_finditer = infix_re.finditer
```

**retrieving data**

```python
# URL atualizada para Spring 2025 Ready-to-Wear
url = 'https://www.vogue.com/fashion-shows/fall-2025-ready-to-wear'

# raspagem de links das páginas de shows
response_show = requests.get(url)
list_fashion_shows = []
if response_show.status_code == 200:
    soup_show = BeautifulSoup(response_show.text, 'html.parser')
    elements = soup_show.find_all('a', class_='NavigationInternalLink-cWEaeo')

    for element in elements:
        if 'fashion-shows' in element['href']:
            list_fashion_shows.append(element['href'])

articles = []

# percorrendo cada link e armazenando o conteúdo
for url_fashion_show in list_fashion_shows:
    full_url = 'https://www.vogue.com' + url_fashion_show
    response_show = requests.get(full_url)

    if response_show.status_code == 200:
        soup_show = BeautifulSoup(response_show.text, 'html.parser')
        show_text_element = soup_show.find('div', class_='body__inner-container')

        try:
            articles.append(show_text_element.text)
            print(f'Sucesso: {full_url}')
        except:
            print(f'Erro: {full_url}')

```

```python
# salvando os artigos em um arquivo .csv
df_articles = pd.DataFrame({'article': articles})
df_articles.to_csv("fall_2025.csv", index=False)
```

# **pré-processsamento de texto**

pré-processamento de texto

```python
# função de limpeza de texto
def clean_text(text):
    text = unicodedata.normalize("NFKD", text).encode("ascii", "ignore").decode("utf-8", "ignore")
    text = text.lower()
    text = re.sub(r"\d+", "", text)
    text = re.sub(r"[^\w\s-]", "", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

# garantir a remoção de stopwords manualmente
stopwords_spacy = nlp.Defaults.stop_words

def lemmatize_article(text):
    doc = nlp(text)
    return [
        token.lemma_ if token.lemma_ != "-PRON-" else token.text
        for token in doc
        if token.text.lower() not in stopwords_spacy  # Remove explicitamente stopwords
        and not token.is_punct  # Remove pontuação
    ]



# combinar lemmas em strings para n-gramas
def combine_lemmas(preprocessed_articles):
    return [" ".join(article) for article in preprocessed_articles]

# pipeline completo para processamento de artigos
def preprocess_articles(articles):
    cleaned_articles = [clean_text(article) for article in articles]
    lemmatized_articles = [lemmatize_article(article) for article in cleaned_articles]
    combined_articles = combine_lemmas(lemmatized_articles)
    return combined_articles

# criar duas versões do corpus:
preprocessed_articles_list = preprocess_articles(articles)  # Lista de strings (para n-gramas)
preprocessed_articles_str = " ".join(preprocessed_articles_list)  # String única (para busca)


```

```python
# n-gramas gerais

# função para gerar n-gramas e exibir
def generate_ngrams(corpus, n, top_k=20, filter_fashion=False):
    # Força o CountVectorizer a usar os tokens já separados (split por espaço)
    vectorizer = CountVectorizer(
        ngram_range=(n, n),
        tokenizer=lambda text: text.split(),
        preprocessor=lambda text: text
    )
    X = vectorizer.fit_transform(corpus)
    ngram_counts = Counter(dict(zip(vectorizer.get_feature_names_out(), X.sum(axis=0).A1)))

    if filter_fashion:
        fashion_set = set(word for category in fashion_categories.values() for word in category)
        ngram_counts = {
            ngram: count
            for ngram, count in ngram_counts.items()
            if any(word in fashion_set for word in ngram.split())
        }

    return Counter(ngram_counts).most_common(top_k)


def plot_ngrams_color(ngrams, title, color):
    """
    Plota os n-gramas usando uma cor específica.
    - `ngrams`: lista de tuplas (n-grama, frequência)
    - `title`: título do gráfico
    - `color`: cor para as barras (Neon Lime Green para gerais, Neon Pink para moda)
    """
    # converter os n-gramas para um df
    df = pd.DataFrame(ngrams, columns=["N-Grama", "Frequência"])

    plt.figure(figsize=(10, 6))
    plt.barh(df["N-Grama"], df["Frequência"], color=color)
    plt.xlabel("Frequência", fontsize=14)
    plt.ylabel("N-Gramas", fontsize=14)
    plt.title(title, fontsize=16)
    plt.gca().invert_yaxis()
    plt.show()

# generating ngrams with the function generate_ngrams
general_bigrams = generate_ngrams(preprocessed_articles_list, 2, top_k=20)
fashion_bigrams = generate_ngrams(preprocessed_articles_list, 2, top_k=20, filter_fashion=True)
general_trigrams = generate_ngrams(preprocessed_articles_list, 3, top_k=30)
fashion_trigrams = generate_ngrams(preprocessed_articles_list, 3, top_k=30, filter_fashion=True)

# exibir gráficos com cores distintas:
plot_ngrams_color(general_bigrams, "Top 20 Bigramas Gerais", "#C0FF00")
plot_ngrams_color(fashion_bigrams, "Top 20 Bigramas de Moda", "#FF007F")
plot_ngrams_color(fashion_trigrams, "Top 30 Trigramas de Moda", "#FF007F")
plot_ngrams_color(general_trigrams, "Top 30 Trigramas Gerais", "#C0FF00")
```

nuvem de palavras

```python
# criar nuvem de palavras para cada categoria baseada no corpus real
def criar_nuvem_categoria(categoria, palavras_categorias, corpus):
    # Transformar corpus em uma única string e dividir em palavras
    todas_palavras = " ".join(corpus).split()

    # filtrar apenas as palavras da categoria que realmente aparecem no corpus
    palavras_frequentes = [word for word in todas_palavras if word in palavras_categorias]

    # contar a frequência real dessas palavras no corpus
    contagem = Counter(palavras_frequentes)

    # se não houver palavras suficientes, pular a categoria
    if not contagem:
        print(f"⚠️ Nenhuma palavra da categoria '{categoria}' encontrada no corpus. Pulando...")
        return

    # gerar a nuvem de palavras
    wordcloud = WordCloud(
        width=800,
        height=400,
        background_color="white",
        max_words=100,
        color_func=color_function
    ).generate_from_frequencies(contagem)

    # exibir a nuvem de palavras
    plt.figure(figsize=(10, 5))
    plt.imshow(wordcloud, interpolation="bilinear")
    plt.axis("off")
    plt.title(f"Nuvem de Palavras: {categoria}", fontsize=14, fontweight="bold", color="black")
    plt.show()

# gerar nuvem de palavras para cada categoria baseada no corpus real
for categoria, palavras in fashion_categories.items():
    criar_nuvem_categoria(categoria, palavras, preprocessed_articles_list)
```

```python
# função para gerar bigramas a partir do corpus
def gerar_bigramas(corpus):
    """Gera bigramas (pares de palavras consecutivas) a partir do corpus processado."""
    bigramas = []
    for texto in corpus:
        palavras = texto.split()  # Divide cada texto em palavras
        bigramas.extend([" ".join(pair) for pair in zip(palavras, palavras[1:])])  # Cria pares consecutivos
    return bigramas

# criar nuvem de palavras para cada categoria baseada nos bigramas do corpus real
def criar_nuvem_bigramas(categoria, palavras_categorias, bigramas_corpus):
    """
    Gera e exibe uma nuvem de palavras baseada em bigramas filtrados por categoria,
    e inclui uma checagem de frequência antes da visualização.
    """
    # filtrar bigramas que contêm pelo menos uma palavra da categoria
    bigramas_filtrados = [
        bigrama for bigrama in bigramas_corpus
        if any(word in palavras_categorias for word in bigrama.split())
    ]

    # contar a frequência real desses bigramas
    contagem = Counter(bigramas_filtrados)

    # checagem de frequência antes da nuvem
    if contagem:
        print(f"\n📊 Frequências dos Top 10 Bigramas na categoria: {categoria}")
        print(contagem.most_common(10))  # Exibe os 10 bigramas mais comuns para verificação
    else:
        print(f"⚠️ Nenhum bigrama relevante encontrado na categoria '{categoria}'. Pulando...")
        return

    # gerar a nuvem de palavras
    wordcloud = WordCloud(
        width=800,
        height=400,
        background_color="white",
        max_words=100,
        color_func=color_function
    ).generate_from_frequencies(contagem)

    # exibir a nuvem de palavras
    plt.figure(figsize=(10, 10))
    plt.imshow(wordcloud, interpolation="bilinear")
    plt.axis("off")
    plt.title(f"Nuvem de Bigramas: {categoria}", fontsize=14, fontweight="bold", color="black")
    plt.show()

# gerar os bigramas do corpus
bigramas_corpus = gerar_bigramas(preprocessed_articles_list)

# gerar nuvem de palavras para cada categoria baseada nos bigramas reais do corpus
for categoria, palavras in fashion_categories.items():
    criar_nuvem_bigramas(categoria, palavras, bigramas_corpus)
```

```python
def rank_fashion_categories(fashion_categories, bigrams_corpus):
    """
    Conta quantos bigramas de cada categoria aparecem nos desfiles.
    """
    category_counts = {category: 0 for category in fashion_categories.keys()}

    for category, words in fashion_categories.items():
        for bigram in bigrams_corpus:
            if any(word in bigram.split() for word in words):
                category_counts[category] += 1

    # criar df
    df_categories = pd.DataFrame(list(category_counts.items()), columns=["Categoria", "Frequência"])
    df_categories = df_categories.sort_values(by="Frequência", ascending=False)

    # exibir DataFrame no Colab
    from IPython.display import display
    display(df_categories)

    return df_categories

# gerar ranking de categorias
df_category_ranking = rank_fashion_categories(fashion_categories, bigramas_corpus)
```

```python
def search_urls_by_expression(expression, articles, list_fashion_shows):
    """
    Pesquisa os artigos que contêm a expressão e retorna as URLs correspondentes.

    Args:
        expression (str): A expressão a ser pesquisada.
        articles (list): Lista de textos dos artigos.
        list_fashion_shows (list): Lista de URLs dos desfiles.

    Returns:
        list: URLs dos artigos que contêm a expressão.
    """
    # Normalizar a expressão para comparação
    expression = expression.lower()

    # Lista para armazenar as URLs dos artigos que contêm a expressão
    matching_urls = []

    # Iterar sobre os artigos e URLs correspondentes
    for i, article in enumerate(articles):
        # Verificar se a expressão está presente no texto
        if expression in article.lower():
            # Adicionar a URL correspondente à lista
            full_url = 'https://www.vogue.com' + list_fashion_shows[i]
            matching_urls.append(full_url)
            print(f"🔗 URL encontrada: {full_url}")

    if not matching_urls:
        print("❌ Nenhuma correspondência encontrada para a expressão.")

    return matching_urls

# Exemplo de uso
expression = "velvet"  # Insira a expressão que deseja buscar
matching_urls = search_urls_by_expression(expression, articles, list_fashion_shows)
print(f"\n🌐 URLs que contêm a expressão '{expression}': {matching_urls}")
```
